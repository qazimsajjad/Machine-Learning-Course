## Haar Cascade

* Haar Cascade is an object detection method used to identify objects such as faces, eyes, and other features in images. 
* It is based on Haar-like features, which are simple rectangular patterns used to detect edges, lines, and textures. 
* The method uses a cascade of classifiers trained with positive and negative images to recognize specific objects efficiently.
* The Haar cascade algorithm was developed by Paul Viola and Michael Jones in 2001 in their paper titled `"Rapid Object Detection using a Boosted Cascade of Simple Features."`
* The method was designed for fast face detection and became widely used due to its efficiency in real-time applications.
* It is a conventional machine learning-based object detection method. 




### Different Haar Cascades for Face, Eye, etc.
Haar cascades are pre-trained classifiers available in OpenCV for detecting different objects. Some commonly used Haar cascade classifiers include:

* **haarcascade_frontalface_default.xml** → Detects frontal faces
* **haarcascade_frontalface_alt.xml** → Alternative frontal face detection
* **haarcascade_eye.xml** → Detects eyes
* **haarcascade_eye_tree_eyeglasses.xml** → Detects eyes even with glasses
* **haarcascade_smile.xml** → Detects smiles
* **haarcascade_fullbody.xml** → Detects full body
* **haarcascade_car.xml** → Detects cars


## Face Detection

```python
import cv2
import matplotlib.pyplot as plt
```

```python
def imshow(image, title=None, size=5):
    imgType = image.shape
    plt.figure(figsize=(size, size))
    if len(imgType)>2:
        plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB), 'gray')
    else:
        plt.imshow(image, 'gray')
    plt.title(title)
    plt.axis('off')
    plt.show()
```

```python
image = cv2.imread('./Trump.jpg')
imshow(image, 'InputImage')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cascade = cv2.CascadeClassifier('./haarcascade_frontalface_default.xml')
face = cascade.detectMultiScale(gray, scaleFactor=1.05, minNeighbors=5,
                                minSize=(30, 30))
for x,y, w,h in face:
    cv2.rectangle(image, (x,y),(x+w, y+h), (0,0,255), 2)
imshow(image, 'FaceImage')
```
